### 1. Install and Import Libraries

Here I am installing the necessary libraries to create the basic graph with nodes and edges. I am using networkx and matplotlib; networkx allows me to create nodes and edges while mathplotlib allows us to produce the plot.

```python
#!pip install networkx matplotlib
import networkx as nx
import matplotlib.pyplot as plt
```

### 2. Creating the graph

In this section, I am creating an empty graph using networkx in which I can  fill with a list of people's name, to then add them to the graph as nodes. Next, I create a list of relationships between the people. Each relationship represents an edge connecting two nodes. Finally, I add these edges to the graph.

```python
# Create Graph
hello_graph = nx.Graph()

# Add nodes
team = ["Andre", "Carol", "Fernando", "Beverly", "Diane", "Garth", "Ed", "Heather", "Ike", "Jane"]

hello_graph.add_nodes_from(team)

# Add edges
edges = [
    ("Andre", "Carol"),
    ("Carol", "Fernando"),
    ("Andre", "Fernando"),
    ("Andre", "Diane"),
    ("Diane", "Fernando"),
    ("Andre", "Beverly"),
    ("Beverly", "Diane"),
    ("Beverly", "Garth"),
    ("Diane", "Carol"),
    ("Diane", "Garth"),
    ("Diane", "Ed"),
    ("Beverly", "Ed"),
    ("Ed", "Garth"),
    ("Fernando", "Garth"),
    ("Fernando", "Heather"),
    ("Garth", "Heather"),
    ("Heather", "Ike"),
    ("Ike", "Jane")]

hello_graph.add_edges_from(edges)
```

### 3. Positioning and plotting

Here, I am once again using a networkx layout to automatically position the nodes in the graph for me with seed = 100, to ensure replication. Then I am using nx.draw to create the graph with the colored nodes and edges, as well as labels.

```python

# Position nodes
pos = nx.spring_layout(hello_graph, seed=42)

# Re-size plot/graph
plt.figure(figsize=(12, 8))

# Graph
nx.draw(
    hello_graph,
    pos,
    with_labels=True,
    node_color="pink",
    node_size=2600,
    font_size=10,
    edge_color="indigo",
    width=3)

plt.title("Hello, Graph World")
plt.axis("off")
plt.show()
```
